```python
import pandas as pd
import sportsdataverse as sdv
```

```python
# Grabbing data from the 2023-24 season
SEASON = 2024 

schedule = sdv.nba.load_nba_schedule(seasons=[SEASON]).to_pandas()
team_box = sdv.nba.load_nba_team_boxscore(seasons=[SEASON]).to_pandas()
```

The `season_type` of 2 represents regular season games 

## Creating the master dataset


```python
# Filtering rows and selecting columns
game_cols = ['game_id', 'game_date', 'season', 'home_id',
            'away_id', 'home_abbreviation', 'away_abbreviation',
            'home_score', 'away_score']

games = schedule.loc[
    (schedule['season_type'] == 2) & 
    (schedule['status_type_name'] == 'STATUS_FINAL') &
    (schedule['status_type_completed'] == True),
    game_cols
].copy()

games = games.rename(columns={
    'home_abbreviation': 'home_team',
    'away_abbreviation': 'away_team'
})

games = games.dropna()
games['game_date'] = pd.to_datetime(games['game_date'])
games['home_win'] = (games['home_score'] > games['away_score']).astype(int)
games = games.sort_values(['game_date', 'game_id']).reset_index(drop=True)

display(games.head())
```

## Validating the output

```python
print("Number of games:", len(games))
print("Unique games:", games["game_id"].nunique())
print("Home win percentage:", games["home_win"].mean())

assert games["game_id"].is_unique
assert games["home_win"].isin([0, 1]).all()
assert not (games["home_score"] == games["away_score"]).any()
```

### Saving the results

```python
from pathlib import Path

ROOT = Path.cwd()

if ROOT.name == 'notebooks':
    ROOT = ROOT.parent

raw_path = ROOT / 'data' / 'raw'
processed_path = ROOT / 'data' / 'processed' 

raw_path.mkdir(parents=True, exist_ok=True)
processed_path.mkdir(parents=True, exist_ok=True)

schedule.to_parquet(raw_path / "schedule_2024.parquet")
team_box.to_parquet(raw_path / "team_box_2024.parquet")
games.to_parquet(processed_path / "games_2024.parquet")
```
